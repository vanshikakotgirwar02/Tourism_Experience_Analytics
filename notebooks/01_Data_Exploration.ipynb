```python
# import libraries 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings

warnings.filterwarnings("ignore")

print("Libraries imported successfully!")
```

```python
# Define the location of our dataset folder

DATA_PATH = "../data"

# Display the data folder path
print("Data folder path:", DATA_PATH)
```

```python
# Get the names of all files present in the data folder
data_files = os.listdir(DATA_PATH)

# Display a heading
print("Files available in the data folder:\n")

# Display each file name
for file in data_files:
    print("-", file)
```

```python
# Load the User dataset
user_df = pd.read_excel("../data/User.xlsx")

# Load the Type dataset
type_df = pd.read_excel("../data/Type.xlsx")

# Load the Transaction dataset
transaction_df = pd.read_excel("../data/Transaction.xlsx")

# Load the Region dataset
region_df = pd.read_excel("../data/Region.xlsx")

# Load the Mode dataset
mode_df = pd.read_excel("../data/Mode.xlsx")

# Load the Item dataset
item_df = pd.read_excel("../data/Item.xlsx")

# Load the Country dataset
country_df = pd.read_excel("../data/Country.xlsx")

# Load the Continent dataset
continent_df = pd.read_excel("../data/Continent.xlsx")

# Load the City dataset
city_df = pd.read_excel("../data/City.xlsx")

# Load the additional attraction-site dataset
attraction_df = pd.read_excel("../data/Updated_Item.xlsx")

# Confirm that all datasets have been loaded
print("All datasets loaded successfully!")
```

```python
# Store all datasets in a dictionary

datasets = {
    "User": user_df,
    "Type": type_df,
    "Transaction": transaction_df,
    "Region": region_df,
    "Mode": mode_df,
    "Item": item_df,
    "Country": country_df,
    "Continent": continent_df,
    "City": city_df,
    "Attraction": attraction_df
}

# Display the names of all loaded datasets
print("Loaded datasets:")
for name in datasets:
    print("-", name)
```

```python
# Create a summary table containing the number of rows and columns
dataset_summary = pd.DataFrame({
    "Dataset": list(datasets.keys()),
    
    # Number of rows in each dataset
    "Rows": [df.shape[0] for df in datasets.values()],
    
    # Number of columns in each dataset
    "Columns": [df.shape[1] for df in datasets.values()]
})

# Display the dataset summary
dataset_summary
```

```python
# Display the column names for every dataset

for name, df in datasets.items():
    
    # Print a separator to make the output easier to read
    print("=" * 60)
    
    # Print the dataset name
    print(f"{name} Dataset")
    
    print("=" * 60)
    
    # Display all column names
    print(df.columns.tolist())
    
    print()
```

```python
# Display the first five rows of every dataset

for name, df in datasets.items():
    
    print("=" * 60)
    print(f"{name} Dataset - First 5 Rows")
    print("=" * 60)
    
    # Display the first five records
    display(df.head())
```

```python
# Create a compact summary of all datasets
# This helps us understand the structure without producing a huge output

structure_summary = []

for name, df in datasets.items():
    
    # Store basic information about each dataset
    structure_summary.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Column Names": ", ".join(df.columns)
    })

# Convert the information into a DataFrame
structure_summary_df = pd.DataFrame(structure_summary)

# Display the complete summary
display(structure_summary_df)
```

```python
# Display the data type of every column in each dataset

for name, df in datasets.items():
    
    print("=" * 70)
    print(f"{name} Dataset - Data Types")
    print("=" * 70)
    
    # Create a small table containing column names and their data types
    dtype_table = pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.astype(str)
    })
    
    display(dtype_table)
```

```python
# Check the number of missing values in every column

for name, df in datasets.items():
    
    print("=" * 70)
    print(f"{name} Dataset - Missing Values")
    print("=" * 70)
    
    # Count missing values for each column
    missing = df.isnull().sum()
    
    # Keep only columns that contain missing values
    missing = missing[missing > 0]
    
    if len(missing) == 0:
        print("No missing values found.")
    else:
        display(missing.to_frame("Missing Values"))
```

```python
# Check for duplicate rows in every dataset

duplicate_summary = []

for name, df in datasets.items():
    
    # Count completely duplicated rows
    duplicate_count = df.duplicated().sum()
    
    duplicate_summary.append({
        "Dataset": name,
        "Duplicate Rows": duplicate_count
    })

# Convert results into a DataFrame
duplicate_summary_df = pd.DataFrame(duplicate_summary)

# Display duplicate counts
display(duplicate_summary_df)
```

```python
# Create a summary of missing values for all datasets
# This gives us an overall view of data completeness

missing_summary = []

for name, df in datasets.items():
    
    # Calculate total missing values
    total_missing = df.isnull().sum().sum()
    
    # Calculate total number of cells
    total_cells = df.shape[0] * df.shape[1]
    
    # Calculate missing-value percentage
    missing_percentage = (total_missing / total_cells) * 100
    
    missing_summary.append({
        "Dataset": name,
        "Total Missing Values": total_missing,
        "Missing Percentage": round(missing_percentage, 2)
    })

# Convert the results into a DataFrame
missing_summary_df = pd.DataFrame(missing_summary)

# Display the summary
display(missing_summary_df)
```

# RATING ANALYSIS

```python
# Display basic statistical information about attraction ratings

display(transaction_df["Rating"].describe())
```

```python
# Check which rating values are present in the transaction dataset

unique_ratings = sorted(transaction_df["Rating"].unique())

print("Unique rating values:")
print(unique_ratings)
```

```python
# Count how frequently each rating occurs

rating_counts = (
    transaction_df["Rating"]
    .value_counts()
    .sort_index()
)

display(rating_counts.to_frame("Number of Ratings"))
```

```python
# Visualize the distribution of tourist ratings

plt.figure(figsize=(8, 5))

sns.countplot(
    data=transaction_df,
    x="Rating"
)

plt.title("Distribution of Attraction Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Calculate the overall average rating given by tourists

average_rating = transaction_df["Rating"].mean()

print("Overall Average Rating:", round(average_rating, 2))
```

```python
# Calculate the average rating for each visit mode

rating_by_mode = (
    transaction_df
    .groupby("VisitMode")["Rating"]
    .agg(["count", "mean"])
    .sort_values("mean", ascending=False)
)

rating_by_mode.columns = ["Number of Visits", "Average Rating"]

display(rating_by_mode)
```

```python
# Calculate the average rating for each visit year

rating_by_year = (
    transaction_df
    .groupby("VisitYear")["Rating"]
    .agg(["count", "mean"])
    .sort_index()
)

rating_by_year.columns = ["Number of Visits", "Average Rating"]

display(rating_by_year)
```

```python
# Visualize how average ratings change across different years

plt.figure(figsize=(10, 5))

sns.lineplot(
    data=rating_by_year,
    x=rating_by_year.index,
    y="Average Rating",
    marker="o"
)

plt.title("Average Rating by Year")
plt.xlabel("Visit Year")
plt.ylabel("Average Rating")
plt.tight_layout()
plt.show()
```

```python
# Calculate the average rating for each visit month

rating_by_month = (
    transaction_df
    .groupby("VisitMonth")["Rating"]
    .agg(["count", "mean"])
    .sort_index()
)

rating_by_month.columns = ["Number of Visits", "Average Rating"]

display(rating_by_month)
```

```python
# Visualize the average rating across months

plt.figure(figsize=(10, 5))

sns.lineplot(
    data=rating_by_month,
    x=rating_by_month.index,
    y="Average Rating",
    marker="o"
)

plt.title("Average Rating by Visit Month")
plt.xlabel("Visit Month")
plt.ylabel("Average Rating")
plt.xticks(range(1, 13))
plt.tight_layout()
plt.show()
```

```python
# Calculate the number of ratings and average rating for each attraction

attraction_rating = (
    transaction_df
    .groupby("AttractionId")["Rating"]
    .agg(["count", "mean"])
)

attraction_rating.columns = ["Number of Ratings", "Average Rating"]

# Sort attractions by average rating
top_rated_attractions = attraction_rating.sort_values(
    "Average Rating",
    ascending=False
)

# Display the top 10
display(top_rated_attractions.head(10))
```

```python
# Count how many times each attraction was visited

most_visited = (
    transaction_df["AttractionId"]
    .value_counts()
    .head(10)
)

display(most_visited.to_frame("Number of Visits"))
```

```python
# Compare the distribution of ratings across different visit modes

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=transaction_df,
    x="VisitMode",
    y="Rating"
)

plt.title("Rating Distribution by Visit Mode")
plt.xlabel("Visit Mode")
plt.ylabel("Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# RELATIONSHIP VALIDATION

# Transaction → User validation

```python
# Check whether every UserId in Transaction exists in the User dataset

transaction_users = set(transaction_df["UserId"])
known_users = set(user_df["UserId"])

missing_users = transaction_users - known_users

print("User IDs in transactions:", len(transaction_users))
print("User IDs not found in User dataset:", len(missing_users))
```

# Transaction → Attraction validation

```python
# Check whether every AttractionId in Transaction exists in Attraction data

transaction_attractions = set(transaction_df["AttractionId"])
known_attractions = set(attraction_df["AttractionId"])

missing_attractions = transaction_attractions - known_attractions

print("Attraction IDs in transactions:", len(transaction_attractions))
print("Attraction IDs not found:", len(missing_attractions))
```

# User → City validation

```python
# Check whether User CityIds exist in the City dataset
# Missing CityId values are excluded from this validation

user_city_ids = set(user_df["CityId"].dropna())
known_city_ids = set(city_df["CityId"])

missing_city_ids = user_city_ids - known_city_ids

print("City IDs used by users:", len(user_city_ids))
print("City IDs not found in City dataset:", len(missing_city_ids))
```

# City → Country validation

```python
# Check whether every CountryId used in City exists in Country data

city_country_ids = set(city_df["CountryId"])
known_country_ids = set(country_df["CountryId"])

missing_country_ids = city_country_ids - known_country_ids

print("Country IDs used by cities:", len(city_country_ids))
print("Country IDs not found:", len(missing_country_ids))
```

# Country → Region validation

```python
# Check whether every RegionId used in Country exists in Region data

country_region_ids = set(country_df["RegionId"])
known_region_ids = set(region_df["RegionId"])

missing_region_ids = country_region_ids - known_region_ids

print("Region IDs used by countries:", len(country_region_ids))
print("Region IDs not found:", len(missing_region_ids))
```

# Region → Continent validation

```python
# Check whether every ContinentId used in Region exists in Continent data

region_continent_ids = set(region_df["ContinentId"])
known_continent_ids = set(continent_df["ContinentId"])

missing_continent_ids = region_continent_ids - known_continent_ids

print("Continent IDs used by regions:", len(region_continent_ids))
print("Continent IDs not found:", len(missing_continent_ids))
```

# Attraction → Type validation

```python
# Check whether every AttractionTypeId in Attraction data
# exists in the Type dataset

attraction_type_ids = set(attraction_df["AttractionTypeId"])
known_type_ids = set(type_df["AttractionTypeId"])

missing_type_ids = attraction_type_ids - known_type_ids

print("Attraction type IDs:", len(attraction_type_ids))
print("Attraction type IDs not found:", len(missing_type_ids))
```

# Attraction → City validation

```python
# Check whether every AttractionCityId exists in the City dataset

attraction_city_ids = set(attraction_df["AttractionCityId"])
known_city_ids = set(city_df["CityId"])

missing_attraction_cities = attraction_city_ids - known_city_ids

print("Cities referenced by attractions:", len(attraction_city_ids))
print("Attraction city IDs not found:", len(missing_attraction_cities))
```

# DATE ANALYSIS

# Visit year and month range

```python
# Examine the time period covered by the transaction data

print("Minimum Visit Year:", transaction_df["VisitYear"].min())
print("Maximum Visit Year:", transaction_df["VisitYear"].max())

print("Minimum Visit Month:", transaction_df["VisitMonth"].min())
print("Maximum Visit Month:", transaction_df["VisitMonth"].max())
```

```python
# Calculate the Interquartile Range (IQR) for ratings
# This helps identify statistically unusual rating values

Q1 = transaction_df["Rating"].quantile(0.25)
Q3 = transaction_df["Rating"].quantile(0.75)

IQR = Q3 - Q1

# Calculate lower and upper boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# Identify ratings outside the statistical IQR boundaries

rating_outliers = transaction_df[
    (transaction_df["Rating"] < lower_bound) |
    (transaction_df["Rating"] > upper_bound)
]

print("Number of potential rating outliers:", len(rating_outliers))

display(rating_outliers.head(10))
```
