```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

print("Libraries imported successfully!")
```

```python
# Load the cleaned dataset created during preprocessing

input_file = "../data/tourism_experience_processed.csv"

df = pd.read_csv(input_file)

print("Processed dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
# Display all columns available in the processed dataset

print("Available columns:")

for i, col in enumerate(df.columns, start=1):
    print(i, "->", col)
```

```python
# Display the first 10 records

display(df.head(10))
```

```python
# Calculate the total number of visits made by each user

user_visit_count = (
    df.groupby("UserId")["TransactionId"]
    .transform("count")
)

df["UserVisitCount"] = user_visit_count

display(
    df[
        ["UserId", "TransactionId", "UserVisitCount"]
    ].head(10)
)
```

```python
# Calculate the total number of visits received by each attraction

attraction_visit_count = (
    df.groupby("AttractionId")["TransactionId"]
    .transform("count")
)

df["AttractionVisitCount"] = attraction_visit_count

display(
    df[
        ["AttractionId", "Attraction", "AttractionVisitCount"]
    ].head(10)
)
```

```python
# Calculate the average rating for each attraction

attraction_avg_rating = (
    df.groupby("AttractionId")["Rating"]
    .transform("mean")
)

df["AttractionAverageRating"] = attraction_avg_rating

display(
    df[
        [
            "AttractionId",
            "Attraction",
            "Rating",
            "AttractionAverageRating"
        ]
    ].head(10)
)
```

```python
# Create a popularity category based on attraction visit count

def classify_popularity(visits):

    if visits >= 100:
        return "Highly Popular"

    elif visits >= 50:
        return "Popular"

    else:
        return "Less Popular"


df["AttractionPopularity"] = df[
    "AttractionVisitCount"
].apply(classify_popularity)

display(
    df[
        [
            "Attraction",
            "AttractionVisitCount",
            "AttractionPopularity"
        ]
    ].head(10)
)
```

```python
# Normalize the rating from 1-5 into a 0-1 scale

df["RatingScore"] = df["Rating"] / 5

display(
    df[
        ["Rating", "RatingScore"]
    ].head(10)
)
```

```python
# Group visits into broad time periods

df["VisitYearGroup"] = pd.cut(
    df["VisitYear"],
    bins=[-np.inf, 2010, 2015, 2020, np.inf],
    labels=[
        "Before 2010",
        "2010-2015",
        "2016-2020",
        "After 2020"
    ]
)

display(
    df[
        ["VisitYear", "VisitYearGroup"]
    ].head(10)
)
```

```python
# Calculate the average rating given by each user

df["UserAverageRating"] = (
    df.groupby("UserId")["Rating"]
    .transform("mean")
)

display(
    df[
        [
            "UserId",
            "Rating",
            "UserAverageRating"
        ]
    ].head(10)
)
```

```python
# Calculate the total number of attraction visits in each city

df["CityVisitCount"] = (
    df.groupby("CityId")["TransactionId"]
    .transform("count")
)

display(
    df[
        [
            "CityId",
            "CityName",
            "CityVisitCount"
        ]
    ].head(10)
)
```

```python
# Check the dataset after feature engineering

print("Dataset shape after feature engineering:")
print(df.shape)

print("\nNew features created:")

new_features = [
    "UserVisitCount",
    "AttractionVisitCount",
    "AttractionAverageRating",
    "AttractionPopularity",
    "RatingScore",
    "VisitYearGroup",
    "UserAverageRating",
    "CityVisitCount"
]

for feature in new_features:
    print("-", feature)
```

```python
# Check for missing values after feature engineering

missing_values = df.isnull().sum()

missing_values = missing_values[
    missing_values > 0
].sort_values(ascending=False)

print("Missing values:")

if len(missing_values) == 0:
    print("No missing values found!")
else:
    display(missing_values)
```

```python
# Save the feature-engineered dataset

output_file = "../data/tourism_experience_features.csv"

df.to_csv(
    output_file,
    index=False
)

print("Feature-engineered dataset saved successfully!")
print("File:", output_file)
```

```python
# Verify that the feature-engineered file exists

if os.path.exists(output_file):

    file_size = os.path.getsize(output_file) / (1024 * 1024)

    print("File created successfully!")
    print("File size:", round(file_size, 2), "MB")

else:

    print("File was not created.")
```

```python
# Final Feature Engineering Summary

print("   FEATURE ENGINEERING COMPLETED")


print("\nFinal dataset shape:", df.shape)

print("\nTotal users:", df["UserId"].nunique())

print("Total attractions:", df["AttractionId"].nunique())

print("Total cities:", df["CityId"].nunique())

print("Total countries:", df["CountryId"].nunique())

print("\nNew features created:")

for feature in new_features:
    print("-", feature)

print("\nSaved file:")
print(output_file)
```
