```python
# Import pandas for data manipulation and analysis
import pandas as pd

# Import numpy for numerical operations
import numpy as np

# Import os for file and folder operations
import os

# Ignore unnecessary warning messages
import warnings
warnings.filterwarnings("ignore")

# Confirm successful import
print("Libraries imported successfully!")
```

```python
# Define the path to the data folder
DATA_PATH = "../data"

# Define the path where the processed dataset will be saved
PROCESSED_PATH = "../data"

# Display the paths
print("Data path:", DATA_PATH)
print("Processed data path:", PROCESSED_PATH)
```

```python
# Load all tourism datasets

user_df = pd.read_excel("../data/User.xlsx")
type_df = pd.read_excel("../data/Type.xlsx")
transaction_df = pd.read_excel("../data/Transaction.xlsx")
region_df = pd.read_excel("../data/Region.xlsx")
mode_df = pd.read_excel("../data/Mode.xlsx")
item_df = pd.read_excel("../data/Item.xlsx")
country_df = pd.read_excel("../data/Country.xlsx")
continent_df = pd.read_excel("../data/Continent.xlsx")
city_df = pd.read_excel("../data/City.xlsx")

# Load the additional attraction-site dataset
attraction_df = pd.read_excel("../data/Updated_Item.xlsx")

# Confirm successful loading
print("All datasets loaded successfully!")
```

```python
# Display the number of rows and columns in each dataset

print("User:", user_df.shape)
print("Type:", type_df.shape)
print("Transaction:", transaction_df.shape)
print("Region:", region_df.shape)
print("Mode:", mode_df.shape)
print("Item:", item_df.shape)
print("Country:", country_df.shape)
print("Continent:", continent_df.shape)
print("City:", city_df.shape)
print("Attraction:", attraction_df.shape)
```

```python
# Display users whose CityId is missing

missing_city_users = user_df[user_df["CityId"].isna()]

display(missing_city_users)
```

```python
# Display the city record where CityName is missing

missing_city_record = city_df[city_df["CityName"].isna()]

display(missing_city_record)
```

```python
# Check and standardize ID columns only when they contain numeric IDs
# We will NOT force AttractionTypeId in the additional attraction file
# to integer because our EDA showed that this column contains text values.

# User dataset
user_df["UserId"] = pd.to_numeric(user_df["UserId"], errors="coerce").astype("Int64")
user_df["ContinentId"] = pd.to_numeric(user_df["ContinentId"], errors="coerce").astype("Int64")
user_df["RegionId"] = pd.to_numeric(user_df["RegionId"], errors="coerce").astype("Int64")
user_df["CountryId"] = pd.to_numeric(user_df["CountryId"], errors="coerce").astype("Int64")
user_df["CityId"] = pd.to_numeric(user_df["CityId"], errors="coerce").astype("Int64")

# Transaction dataset
transaction_df["TransactionId"] = pd.to_numeric(
    transaction_df["TransactionId"], errors="coerce"
).astype("Int64")

transaction_df["UserId"] = pd.to_numeric(
    transaction_df["UserId"], errors="coerce"
).astype("Int64")

transaction_df["AttractionId"] = pd.to_numeric(
    transaction_df["AttractionId"], errors="coerce"
).astype("Int64")

# Type dataset
type_df["AttractionTypeId"] = pd.to_numeric(
    type_df["AttractionTypeId"], errors="coerce"
).astype("Int64")

# Region dataset
region_df["RegionId"] = pd.to_numeric(
    region_df["RegionId"], errors="coerce"
).astype("Int64")

region_df["ContinentId"] = pd.to_numeric(
    region_df["ContinentId"], errors="coerce"
).astype("Int64")

# Mode dataset
mode_df["VisitModeId"] = pd.to_numeric(
    mode_df["VisitModeId"], errors="coerce"
).astype("Int64")

# Item dataset
item_df["AttractionId"] = pd.to_numeric(
    item_df["AttractionId"], errors="coerce"
).astype("Int64")

item_df["AttractionCityId"] = pd.to_numeric(
    item_df["AttractionCityId"], errors="coerce"
).astype("Int64")

item_df["AttractionTypeId"] = pd.to_numeric(
    item_df["AttractionTypeId"], errors="coerce"
).astype("Int64")

# Country dataset
country_df["CountryId"] = pd.to_numeric(
    country_df["CountryId"], errors="coerce"
).astype("Int64")

country_df["RegionId"] = pd.to_numeric(
    country_df["RegionId"], errors="coerce"
).astype("Int64")

# Continent dataset
continent_df["ContinentId"] = pd.to_numeric(
    continent_df["ContinentId"], errors="coerce"
).astype("Int64")

# City dataset
city_df["CityId"] = pd.to_numeric(
    city_df["CityId"], errors="coerce"
).astype("Int64")

city_df["CountryId"] = pd.to_numeric(
    city_df["CountryId"], errors="coerce"
).astype("Int64")

# Additional attraction-site dataset
# Keep AttractionTypeId unchanged for now because it contains text
# such as "Temple" and therefore is not a numeric ID in this file.

attraction_df["AttractionId"] = pd.to_numeric(
    attraction_df["AttractionId"], errors="coerce"
).astype("Int64")

attraction_df["AttractionCityId"] = pd.to_numeric(
    attraction_df["AttractionCityId"], errors="coerce"
).astype("Int64")

print("ID columns standardized safely!")
```

```python
# Verify the important ID columns after conversion

print("User CityId:", user_df["CityId"].dtype)
print("Transaction UserId:", transaction_df["UserId"].dtype)
print("Transaction AttractionId:", transaction_df["AttractionId"].dtype)
print("Attraction TypeId:", attraction_df["AttractionTypeId"].dtype)
print("City CityId:", city_df["CityId"].dtype)
```

```python
# Start the master dataset with transaction records

master_df = transaction_df.copy()

# Display the initial shape
print("Initial master dataset shape:", master_df.shape)
```

```python
# Merge transaction data with user information
# UserId is the common key between the two datasets

master_df = master_df.merge(
    user_df,
    on="UserId",
    how="left"
)

# Display the updated shape
print("After merging User:", master_df.shape)
```

```python
# Merge attraction information using AttractionId

master_df = master_df.merge(
    attraction_df,
    on="AttractionId",
    how="left"
)

# Display the updated shape
print("After merging Attraction:", master_df.shape)
```

```python
# Merge attraction type information
# AttractionTypeId connects Attraction with Type

master_df = master_df.merge(
    type_df,
    on="AttractionTypeId",
    how="left"
)

print("After merging Attraction Type:", master_df.shape)
```

```python
# Select only the required columns from the Mode dataset
# This prevents duplicate VisitMode columns after merging

mode_lookup = mode_df[
    ["VisitModeId", "VisitMode"]
].copy()

# Rename the descriptive column so it is clear in the final dataset
mode_lookup = mode_lookup.rename(
    columns={"VisitMode": "VisitModeName"}
)

# Merge transaction data with visit mode information
# Transaction.VisitMode → Mode.VisitModeId

master_df = master_df.merge(
    mode_lookup,
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left"
)

# Display the updated shape
print("After merging Visit Mode:", master_df.shape)

# Display the mapping to verify the merge
display(
    master_df[
        ["VisitMode", "VisitModeId", "VisitModeName"]
    ].head(10)
)
```

```python
# Check whether all VisitMode values found a matching VisitModeName

missing_visit_modes = master_df["VisitModeName"].isna().sum()

print("Records without a matching Visit Mode:", missing_visit_modes)

display(
    master_df[
        ["VisitMode", "VisitModeId", "VisitModeName"]
    ].head(10)
)
```

# Merge City Information

# AttractionCityId → CityId → CityName

```python
# Select required columns from City dataset

city_lookup = city_df[
    ["CityId", "CityName"]
].drop_duplicates("CityId")

# Merge City information using AttractionCityId

master_df = master_df.merge(
    city_lookup,
    left_on="AttractionCityId",
    right_on="CityId",
    how="left",
    suffixes=("", "_City"),
    validate="many_to_one"
)

# Remove duplicate CityId created by the merge

master_df = master_df.drop(columns=["CityId_City"])

print("After merging City:", master_df.shape)

display(
    master_df[
        ["AttractionCityId", "CityId", "CityName"]
    ].head(10)
)
```

```python
# Check whether all attractions found a matching city

missing_city = master_df["CityName"].isna().sum()

print("Records without a matching City:", missing_city)
```

```python
# Select required columns from Country dataset

country_lookup = country_df[
    ["CountryId", "Country", "RegionId"]
].drop_duplicates("CountryId")

# Merge Country information

master_df = master_df.merge(
    country_lookup,
    on="CountryId",
    how="left",
    suffixes=("", "_Country"),
    validate="many_to_one"
)

print("After merging Country:", master_df.shape)

display(
    master_df[
        ["CountryId", "Country"]
    ].head(10)
)
```

```python
# Check whether all records found a matching country

missing_country = master_df["Country"].isna().sum()

print("Records without a matching Country:", missing_country)
```

```python
# Replace the original RegionId with the RegionId
# obtained from the Country dataset

if "RegionId_Country" in master_df.columns:
    master_df["RegionId"] = master_df["RegionId_Country"]
    master_df = master_df.drop(columns=["RegionId_Country"])

print("RegionId updated successfully!")

display(
    master_df[
        ["CountryId", "Country", "RegionId"]
    ].head(10)
)
```

```python
# Select required columns from Region dataset

region_lookup = region_df[
    ["RegionId", "Region", "ContinentId"]
].drop_duplicates("RegionId")

# Merge Region information

master_df = master_df.merge(
    region_lookup,
    on="RegionId",
    how="left",
    suffixes=("", "_Region"),
    validate="many_to_one"
)

print("After merging Region:", master_df.shape)

display(
    master_df[
        ["RegionId", "Region"]
    ].head(10)
)
```

```python
# Check whether all records found a matching region

missing_region = master_df["Region"].isna().sum()

print("Records without a matching Region:", missing_region)
```

```python
# Replace the original ContinentId with the
# ContinentId obtained from the Region dataset

if "ContinentId_Region" in master_df.columns:
    master_df["ContinentId"] = master_df["ContinentId_Region"]
    master_df = master_df.drop(columns=["ContinentId_Region"])

print("ContinentId updated successfully!")

display(
    master_df[
        ["RegionId", "Region", "ContinentId"]
    ].head(10)
)
```

```python
# Select required columns from Continent dataset

continent_lookup = continent_df[
    ["ContinentId", "Continent"]
].drop_duplicates("ContinentId")

# Merge Continent information

master_df = master_df.merge(
    continent_lookup,
    on="ContinentId",
    how="left",
    validate="many_to_one"
)

print("After merging Continent:", master_df.shape)

display(
    master_df[
        ["ContinentId", "Continent"]
    ].head(10)
)
```

```python
# Check whether all records found a matching continent

missing_continent = master_df["Continent"].isna().sum()

print("Records without a matching Continent:", missing_continent)
```

```python
# Check the final structure after all major merges

print("Final merged dataset shape:", master_df.shape)

print("\nColumns:")
print(master_df.columns.tolist())

display(master_df.head(10))
```

```python
# Check missing values in the merged dataset

missing_summary = master_df.isnull().sum()

missing_summary = missing_summary[
    missing_summary > 0
].sort_values(ascending=False)

print("Columns with missing values:")

display(missing_summary)
```

```python
# Display rows containing missing values

missing_rows = master_df[
    master_df.isnull().any(axis=1)
]

print("Number of rows containing missing values:", len(missing_rows))

display(missing_rows.head(10))
```

```python
# Check remaining missing values

missing_summary = master_df.isnull().sum()

missing_summary = missing_summary[
    missing_summary > 0
].sort_values(ascending=False)

print("Columns with missing values:")

display(missing_summary)
```

```python
# Remove rows where essential tourism information is missing

before_rows = len(master_df)

master_df = master_df.dropna(
    subset=[
        "AttractionId",
        "AttractionCityId",
        "CityName",
        "Country",
        "Region",
        "Continent"
    ]
)

after_rows = len(master_df)

print("Rows before removing missing essential values:", before_rows)
print("Rows after removing missing essential values:", after_rows)
print("Rows removed:", before_rows - after_rows)
```

```python
# Verify remaining missing values

remaining_missing = master_df.isnull().sum()

remaining_missing = remaining_missing[
    remaining_missing > 0
].sort_values(ascending=False)

print("Remaining missing values:")

display(remaining_missing)
```

```python
# Check duplicate Attraction Type columns

if "AttractionType_x" in master_df.columns and "AttractionType_y" in master_df.columns:

    same_percentage = (
        master_df["AttractionType_x"].astype(str)
        ==
        master_df["AttractionType_y"].astype(str)
    ).mean() * 100

    print(
        f"AttractionType columns have {same_percentage:.2f}% matching values."
    )
```

```python
# Keep one AttractionType column

master_df = master_df.drop(
    columns=["AttractionType_y"]
)

master_df = master_df.rename(
    columns={
        "AttractionType_x": "AttractionType"
    }
)

print("Duplicate AttractionType column removed.")

display(
    master_df[
        [
            "AttractionId",
            "AttractionTypeId",
            "Attraction",
            "AttractionType"
        ]
    ].head(10)
)
```

```python
# Keep the descriptive VisitMode name
# and remove the duplicate numeric VisitMode column

master_df = master_df.drop(
    columns=["VisitMode"]
)

master_df = master_df.rename(
    columns={
        "VisitModeId": "VisitModeID",
        "VisitModeName": "VisitMode"
    }
)

print("Visit Mode columns cleaned.")

display(
    master_df[
        ["VisitModeID", "VisitMode"]
    ].head(10)
)
```

```python
# Create a proper VisitDate column

master_df["VisitDate"] = pd.to_datetime(
    master_df["VisitYear"].astype(str)
    + "-"
    + master_df["VisitMonth"].astype(str)
    + "-01",
    errors="coerce"
)

display(
    master_df[
        ["VisitYear", "VisitMonth", "VisitDate"]
    ].head(10)
)
```

```python
# Create rating categories

def classify_rating(rating):

    if rating <= 2:
        return "Low"

    elif rating == 3:
        return "Average"

    else:
        return "High"


master_df["RatingCategory"] = master_df["Rating"].apply(
    classify_rating
)

display(
    master_df[
        ["Rating", "RatingCategory"]
    ].head(10)
)
```

```python
# Create Year-Month column for time-based analysis

master_df["VisitYearMonth"] = (
    master_df["VisitDate"]
    .dt.to_period("M")
    .astype(str)
)

display(
    master_df[
        ["VisitDate", "VisitYearMonth"]
    ].head(10)
)
```

```python
# Check duplicate records in the final dataset

duplicate_count = master_df.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

```python
# Check final data types

print("Final data types:")

display(
    master_df.dtypes.to_frame("Data Type")
)
```

```python
# Final missing-value check

final_missing = master_df.isnull().sum()

final_missing = final_missing[
    final_missing > 0
].sort_values(ascending=False)

print("Final missing values:")

if len(final_missing) == 0:
    print("No missing values remaining!")
else:
    display(final_missing)
```

```python
# Preview the cleaned dataset

display(
    master_df.head(10)
)
```

```python
# Check final dataset dimensions

print("Final number of rows:", master_df.shape[0])
print("Final number of columns:", master_df.shape[1])
```

```python
# Save the processed dataset

output_file = "../data/tourism_experience_processed.csv"

master_df.to_csv(
    output_file,
    index=False
)

print("Processed dataset saved successfully!")
print("File:", output_file)
```
